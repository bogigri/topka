### 26. Свойства логических операций

Программно проверьте на всех комбинациях `True`/`False` свойства:
* коммутативность: `a and b == b and a`, `a or b == b or a`;
* ассоциативность: `(a and b) and c == a and (b and c)`, `(a or b) or c == a or (b or c)`;
* дистрибутивность: `a and (b or c) == (a and b) or (a and c)`, `a or (b and c) == (a or b) and (a or c)`.

Выведите вывод, какие свойства всегда выполняются. Тема: «Свойства логических операций: ассоциативность, коммутативность, дистрибутивность».

```python
print('коммутативность')
for a in range(2):
  for b in range(2):
      print((a and b) == (b and a))
      print((a or b) == (b or a))
print('ассоциативность')
for a in range(2):
  for b in range(2):
    for c in range(2):
      print(((a and b) and c) == (a and (b and c)))
      print(((a or b) or c) == (a or (b or c)))
print('дистрибутивность')
for a in range(2):
  for b in range(2):
    for c in range(2):
      print((a and (b or c)) == ((a and b) or (a and c)))
      print((a or (b and c)) == ((a or b) and (a or c)))
```

### 27. СДНФ: построение функции по таблице истинности

Дана функция большинства `F(a, b, c)` — она равна 1, если среди `a`, `b`, `c` единиц не меньше двух. Выпишите наборы, где `F = 1`, постройте выражение в виде СДНФ (дизъюнкция конъюнкций) и вычислите его для всех 8 комбинаций, сверив с таблицей.

Подсказка (один из вариантов): `F = (a and b) or (a and c) or (b and c)`. Темы: «Логические операции», «Таблица истинности», «Операторы if/else».

```python
d = [False, True]

for a in d:
    for b in d:
        for c in d:

            # Функция большинства
            F = (a and b) or (a and c) or (b and c)

            print(a, b, c, "=>", F)
```
